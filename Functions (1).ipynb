1. What is the difference between a function and a method in Pyhton?
>* In Python, a function is a block of reusable code defined using def, callable independently. A method is a function associated with an object, typically called on instances of a class. Methods implicitly receive the instance (self) as the first argument, while functions do not.


```python
#Example:-
def add(x, y):       # Function
    return x + y

class Math:
    def add(self, x, y):  # Method
        return x + y

print(add(2, 3))            # Function call
print(Math().add(2, 3))     # Method call
```

2. Explain the concept of function arguments and parameters in Pyhton?
>* In Python, parameters are variables listed in a function’s definition, while arguments are the actual values passed to the function when it's called. Parameters define what kind of input a function expects, and arguments provide that input. Python supports different types of arguments: positional, keyword, default, and variable-length (*args, **kwargs), allowing flexible function usage and improved code readability.


```python
#Example:-
def greet(name, greeting="Hello"):  # 'name' and 'greeting' are parameters
    print(f"{greeting}, {name}!")

greet("Ajay")                    # Positional argument
greet(name="Nayan", greeting="Hi") # Keyword arguments
```

6. Explain the concept of generators in Pyhton and how thay are defined.
>* Generators in Python are special functions that return an iterator using the yield keyword instead of return. They allow you to generate a series of values lazily, one at a time, without storing them in memory. Generators are defined like regular functions but use yield to return values, enabling efficient iteration over large datasets.




```python
#Example:-
def count_up_to(limit):
    count = 1
    while count <= limit:
        yield count  # Yielding the current count
        count += 1

# Using the generator
counter = count_up_to(5)
for number in counter:
    print(number)
```

7. What are the advantages of using generators over regular functions?
>* Generators offer several advantages over regular functions. They are memory-efficient since they generate values one at a time and don't store the entire output in memory.
  * This is especially useful when dealing with large datasets or infinite sequences. Generators also enable lazy evaluation, meaning values are computed only when needed, improving performance. Additionally, they provide cleaner and more readable code when handling iterative processes, avoiding the need for manual state management in loops or custom iterators.

```python
#Example:-

def generate_numbers_list(limit):
    return [i for i in range(1, limit + 1)]

def generate_numbers_generator(limit):
    for i in range(1, limit + 1):
        yield i

numbers_list = generate_numbers_list(1000000)  # Creates a large list
numbers_generator = generate_numbers_generator(1000000)  # Generates values one by one

# The generator only uses memory for the current value being processed
print(next(numbers_generator))  # Output: 1
print(next(numbers_generator))  # Output: 2
```

9. Explain the purpose and usage of the 'map()' function in Pyhton.
>* The map() function in Python applies a given function to all items in an iterable (like a list) and returns an iterator.
* It's used to transform each element of the iterable in a concise way. Instead of using a loop, map() allows you to apply a function to every element in a more functional programming style. It’s often used for transformations or calculations.

```python
#Example:-

def square(x):
    return x * x


numbers = [1, 2, 3, 4]


squared_numbers = map(square, numbers)


print(list(squared_numbers))
```

10. What is the difference between 'map()',reduce()'and filter()' functions in Pyhton.

>* The differences between among these three are as follows:-

* map()

i)Applies a function to each item in an iterable.

ii)Returns a new iterator with the results.

iii)Used for transforming data.

* filter()

i)Applies a function that returns True or False to each item.

ii)Keeps only items where the function returns True.

iii)Used for filtering data.

* reduce() (from functools module)

i)Applies a function to pairs of items, reducing the iterable to a single value.

ii)Used for cumulative operations like sum or product.

iii)Returns a single result, not an iterator.

```python
#Example:-

from functools import reduce

numbers = [1, 2, 3, 4, 5]

# map(): double each number
doubled = map(lambda x: x * 2, numbers)
print(list(doubled))

# filter(): keep only even numbers
evens = filter(lambda x: x % 2 == 0, numbers)
print(list(evens))

# reduce(): sum all numbers
total = reduce(lambda x, y: x + y, numbers)
print(total)
```

```python
#4. Write a Python function that checks if a given number is prime or not from 1 to 200.

def is_prime(n):
    if n <= 1:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False

    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False

    return True

for num in range(1, 201):
    if is_prime(num):
        print(num, "is a prime number")

```

```python
#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.count = 0
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration

        if self.count == 0:
            self.count += 1
            return 0
        elif self.count == 1:
            self.count += 1
            return 1
        else:
            next_value = self.a + self.b
            self.a, self.b = self.b, next_value
            self.count += 1
            return next_value

fib = FibonacciIterator(10)

for num in fib:
    print(num, end=" ")

```

```python
#6. Write a generator function in Python that yeilds the powers of 2 up to a given exponent.

def powers_of_two(max_exponent):
    for exp in range(max_exponent + 1):
        yield 2 ** exp

for value in powers_of_two(5):
    print(value)
```

```python
#write a pyhton program,which returns a list with 2-tuples.each tuple consists of the order number and the product of the price per item and the quantity.the product should be increased by10,- $ if the value of the order is smaller than 100,00$.

def calculate_order_totals(orders):

    result = []
    for order in orders:
        order_number, price_per_item, quantity = order
        total = price_per_item * quantity
        if total < 100.0:
            total += 10.0
        result.append((order_number, total))
    return result

# Example usage
orders = [
    (1, 15.0, 5),
    (2, 50.0, 1),
    (3, 20.0, 6),
]
order_totals = calculate_order_totals(orders)
print(order_totals)
```
